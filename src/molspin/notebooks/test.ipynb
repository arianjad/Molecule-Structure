```python
from molspin.core import Basis, QuantumState

b = Basis(
    "HundsCaseB_doublet",
    ("N", "S", "J", "M"),
    [
        QuantumState({"N": 0, "S": 0.5, "J": 0.5, "M": -0.5}),
        QuantumState({"N": 0, "S": 0.5, "J": 0.5, "M":  0.5}),
    ],
)

print(b.qnumbers["M"])   # [-0.5, 0.5]
print(b.qstr_list)       # ('N', 'S', 'J', 'M')
print(b.state(1).values) # {'N': 0, 'S': 0.5, 'J': 0.5, 'M': 0.5}
```
